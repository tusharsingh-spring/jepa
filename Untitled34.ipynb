```python
import os
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import torchvision.datasets as datasets
import torchvision.transforms as transforms
import torchvision.models as models
import matplotlib.pyplot as plt
import random
from tqdm import tqdm
from sentence_transformers import SentenceTransformer

# ==========================================
# 1. DYNAMIC DENSELY CAPTIONED MULTIMODAL DATASET
# ==========================================
CLASS_DENSE_VARIANTS = {
    "AnnualCrop": [
        "Regularly plowed agricultural land with distinct parallel crop rows and soil patterns.",
        "Farmland with visible vegetation grids and tilled earth tracks.",
        "Cultivated crop fields displaying agricultural rows from an aerial perspective."
    ],
    "Forest": [
        "Dense canopied woodland area with heavy green vegetation cover and overlapping tree crowns.",
        "Thick forest terrain filled with natural organic distribution of green trees.",
        "A heavily wooded canopy zone showing dense vegetation patterns."
    ],
    "HerbaceousVegetation": [
        "Open fields covered with wild grass, low bushes, and uncultivated native terrain features.",
        "Grassy meadows and sprawling natural green pastures without human infrastructure.",
        "Low-lying herbaceous plants and open wild shrubbery terrain."
    ],
    "Highway": [
        "Paved transport highway roads with linear asphalt tracks and concrete lanes.",
        "Interstate asphalt corridors cutting across terrain with visible vehicular lanes.",
        "Linear transport highway infrastructure with dark asphalt strips."
    ],
    "Industrial": [
        "Large commercial buildings, metallic warehouse roofs, and concrete parking zones.",
        "Heavy industrial facilities with structured roofing and logistical yards.",
        "Commercial factory complexes and large metallic building structures."
    ],
    "Pasture": [
        "Open green livestock grazing fields, fenced grasslands, and agricultural meadows.",
        "Patchy green pastures used for grazing livestock with clear field boundaries.",
        "Open grassy fields designed for agricultural animal grazing."
    ],
    "PermanentCrop": [
        "Structured orchards, tree plantations, or vineyard rows planted in uniform geometric grids.",
        "Permanently planted tree lines and aligned agricultural crop arrangements.",
        "Grid-aligned orchard trees and structured perennial cultivation plots."
    ],
    "Residential": [
        "Dense suburban layout featuring small roof structures and intersecting neighborhood streets.",
        "Housing blocks and residential properties with interconnected urban paths.",
        "Suburban neighborhood settlement with roof tops and local plot divisions."
    ],
    "River": [
        "A natural flowing water body channel curving through landmasses with visible banks.",
        "Winding river streams flowing through natural terrain boundaries.",
        "An elongated water channel cutting through the landscape."
    ],
    "SeaLake": [
        "A wide open body of water, coastal water body, or inland lake showing smooth surfaces.",
        "Large open water expanses with clear shoreline boundaries.",
        "A expansive static water body showing deep blue or green fluid textures."
    ]
}

class ScaledTrueVLJEPADataset(Dataset):
    def __init__(self, root_dir="./data", max_samples=3000):
        self.transform = transforms.Compose([
            transforms.Resize((224, 224)),
            transforms.ToTensor(),
            transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
        ])
        print("Loading EuroSAT Dataset...")
        self.eurosat = datasets.EuroSAT(root=root_dir, transform=self.transform, download=True)
        self.classes = self.eurosat.classes

        max_samples = min(max_samples, len(self.eurosat))
        indices = list(range(len(self.eurosat)))
        random.seed(42)
        random.shuffle(indices)
        self.indices = indices[:max_samples]

        print("Loading text semantic engine (all-MiniLM-L6-v2)...")
        self.text_engine = SentenceTransformer('all-MiniLM-L6-v2')

    def __len__(self):
        return len(self.indices)

    def __getitem__(self, idx):
        real_idx = self.indices[idx]
        img, label_idx = self.eurosat[real_idx]
        class_name = self.classes[label_idx]

        variants = CLASS_DENSE_VARIANTS.get(class_name, [f"A satellite view of {class_name}."])
        caption = random.choice(variants)

        text_emb = self.text_engine.encode(caption, convert_to_tensor=True)
        text_emb = F.normalize(text_emb, dim=0)

        return img, text_emb, caption


# ==========================================
# 2. NOVEL COUNTERFACTUAL VL-JEPA ARCHITECTURE
# ==========================================
class VisionEncoderX(nn.Module):
    def __init__(self):
        super().__init__()
        self.backbone = models.vit_b_16(weights=models.ViT_B_16_Weights.DEFAULT)
        self.backbone.heads = nn.Identity()

    def forward(self, x):
        return self.backbone(x) # Output dimension: 768


class NovelCounterfactualVLJEPAPredictor(nn.Module):
    """
    Novel Dual-Head Predictor:
    1. Semantic Head: Forecasts exact target text embedding matching.
    2. Delta Head: Computes environmental shift offsets to model terrain transitions.
    """
    def __init__(self, vision_dim=768, query_dim=384, hidden_dim=512, target_dim=384):
        super().__init__()
        self.fusion = nn.Linear(vision_dim + query_dim, hidden_dim)

        self.semantic_head = nn.Sequential(
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Linear(hidden_dim, target_dim)
        )

        self.delta_head = nn.Sequential(
            nn.LayerNorm(hidden_dim),
            nn.GELU(),
            nn.Linear(hidden_dim, target_dim)
        )

    def forward(self, vis_features, query_embeddings):
        combined = torch.cat([vis_features, query_embeddings], dim=-1)
        hidden = self.fusion(combined)

        predicted_latent = F.normalize(self.semantic_head(hidden), dim=-1)
        shift_delta = self.delta_head(hidden) # Spatial/environmental vector drift offset

        return predicted_latent, shift_delta


def apply_random_mask(images, mask_ratio=0.3):
    B, C, H, W = images.shape
    masked_images = images.clone()
    patch_size = 32
    for i in range(0, H, patch_size):
        for j in range(0, W, patch_size):
            if random.random() < mask_ratio:
                masked_images[:, :, i:i+patch_size, j:j+patch_size] = 0.0
    return masked_images


# ==========================================
# 3. QUERY EVALUATION & ADVANCED VISUALIZATION
# ==========================================
def evaluate_novel_vl_jepa(query_text, dataset, vision_encoder, predictor, device, top_k=3):
    vision_encoder.eval()
    predictor.eval()

    query_tensor = dataset.text_engine.encode(query_text, convert_to_tensor=True)
    query_vector = F.normalize(query_tensor, dim=0).to(device)

    dataloader = DataLoader(dataset, batch_size=1, shuffle=False)

    print(f"\n==========================================")
    print(f"NOVEL COUNTERFACTUAL QUERY: '{query_text}'")
    print(f"==========================================")

    similarities = []
    deltas_norm = []
    batch_query_vector = query_vector.unsqueeze(0)

    with torch.no_grad():
        for img, _, _ in dataloader:
            img = img.to(device)
            vis_feats = vision_encoder(img)
            predicted_latent, shift_delta = predictor(vis_feats, batch_query_vector)

            sim = torch.sum(predicted_latent * query_vector, dim=-1)
            similarities.append(sim.cpu())
            deltas_norm.append(torch.norm(shift_delta, dim=-1).cpu())

    similarities = torch.cat(similarities, dim=0)
    deltas_norm = torch.cat(deltas_norm, dim=0)

    top_scores, top_indices = torch.topk(similarities, k=min(top_k, len(dataset)))

    fig, axes = plt.subplots(1, top_k, figsize=(4 * top_k, 4.5))
    if top_k == 1:
        axes = [axes]

    mean = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1)
    std = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)

    for rank, (score, idx) in enumerate(zip(top_scores, top_indices), 1):
        idx_val = idx.item()
        score_val = score.item()
        delta_val = deltas_norm[idx_val].item()

        img_tensor, _, original_caption = dataset[idx_val]
        print(f"  Match {rank}: Index [{idx_val}] | Alignment Score: {score_val:.4f} | Delta Offset: {delta_val:.3f} | Text: {original_caption}")

        unnorm_img = img_tensor * std + mean
        unnorm_img = torch.clamp(unnorm_img, 0, 1)
        img_np = unnorm_img.permute(1, 2, 0).numpy()

        axes[rank-1].imshow(img_np)
        axes[rank-1].set_title(f"Rank {rank}\nScore: {score_val:.3f} | Δ: {delta_val:.2f}", fontsize=9)
        axes[rank-1].axis('off')

    plt.suptitle(f"Counterfactual VL-JEPA Query: '{query_text}'", fontsize=10, fontweight='bold')
    plt.tight_layout()
    plt.show()


# ==========================================
# 4. TRAINING EXECUTION LOOP
# ==========================================
if __name__ == "__main__":
    device = "cuda" if torch.cuda.is_available() else "cpu"
    print(f"Using compute device: {device}")

    train_ds = ScaledTrueVLJEPADataset(root_dir="./data", max_samples=3000)
    dataloader = DataLoader(train_ds, batch_size=64, shuffle=True, num_workers=0)

    vision_encoder = VisionEncoderX().to(device)
    predictor = NovelCounterfactualVLJEPAPredictor(vision_dim=768, query_dim=384, target_dim=384).to(device)

    optimizer = torch.optim.AdamW(
        list(vision_encoder.parameters()) + list(predictor.parameters()),
        lr=1e-4, weight_decay=0.05
    )

    print("\nStarting Counterfactual VL-JEPA Training...")
    vision_encoder.train()
    predictor.train()

    epochs = 3
    for epoch in range(epochs):
        total_loss = 0.0
        for satellite_imgs, target_text_embeds, _ in tqdm(dataloader, desc=f"Epoch {epoch+1}/{epochs}"):
            satellite_imgs = satellite_imgs.to(device)
            target_text_embeds = target_text_embeds.to(device)

            masked_imgs = apply_random_mask(satellite_imgs, mask_ratio=0.3)
            vis_feats = vision_encoder(masked_imgs)

            predicted_latent, shift_delta = predictor(vis_feats, target_text_embeds)

            # Combined Loss: Semantic Cosine Alignment + Regularized Delta Magnitude penalty
            semantic_loss = 1.0 - (predicted_latent * target_text_embeds).sum(dim=-1).mean()
            delta_loss = 0.01 * torch.norm(shift_delta, dim=-1).mean() # Keeps spatial transitions stable

            loss = semantic_loss + delta_loss

            optimizer.zero_grad()
            loss.backward()
            torch.nn.utils.clip_grad_norm_(list(vision_encoder.parameters()) + list(predictor.parameters()), max_norm=1.0)
            optimizer.step()

            total_loss += loss.item()

        print(f"Epoch {epoch+1} | Combined Latent Loss: {total_loss / len(dataloader):.4f}")

    # ==========================================
    # 5. RUN CUSTOM OPEN-VOCABULARY QUERIES
    # ==========================================
    evaluate_novel_vl_jepa("paved transport highway roads with linear asphalt tracks", train_ds, vision_encoder, predictor, device, top_k=3)
    evaluate_novel_vl_jepa("dense canopied woodland area with heavy green vegetation", train_ds, vision_encoder, predictor, device, top_k=3)
    evaluate_novel_vl_jepa("large commercial buildings and concrete parking zones", train_ds, vision_encoder, predictor, device, top_k=3)
```

```python
import os
import random
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import torchvision.datasets as datasets
import torchvision.transforms as transforms
import torchvision.models as models
from torch.utils.checkpoint import checkpoint
from tqdm import tqdm
import matplotlib.pyplot as plt
from sentence_transformers import SentenceTransformer

os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

# ==========================================
# 1. CLASS DENSE CAPTIONS FOR EUROSAT
# ==========================================
CLASS_DENSE_VARIANTS = {
    "AnnualCrop": [
        "Regularly plowed agricultural land with distinct parallel crop rows and soil patterns.",
        "Farmland with visible vegetation grids and tilled earth tracks.",
        "Cultivated crop fields displaying agricultural rows from an aerial perspective."
    ],
    "Forest": [
        "Dense canopied woodland area with heavy green vegetation cover and overlapping tree crowns.",
        "Thick forest terrain filled with natural organic distribution of green trees.",
        "A heavily wooded canopy zone showing dense vegetation patterns."
    ],
    "HerbaceousVegetation": [
        "Open fields covered with wild grass, low bushes, and uncultivated native terrain features.",
        "Grassy meadows and sprawling natural green pastures without human infrastructure.",
        "Low-lying herbaceous plants and open wild shrubbery terrain."
    ],
    "Highway": [
        "Paved transport highway roads with linear asphalt tracks and concrete lanes.",
        "Interstate asphalt corridors cutting across terrain with visible vehicular lanes.",
        "Linear transport highway infrastructure with dark asphalt strips."
    ],
    "Industrial": [
        "Large commercial buildings, metallic warehouse roofs, and concrete parking zones.",
        "Heavy industrial facilities with structured roofing and logistical yards.",
        "Commercial factory complexes and large metallic building structures."
    ],
    "Pasture": [
        "Open green livestock grazing fields, fenced grasslands, and agricultural meadows.",
        "Patchy green pastures used for grazing livestock with clear field boundaries.",
        "Open grassy fields designed for agricultural animal grazing."
    ],
    "PermanentCrop": [
        "Structured orchards, tree plantations, or vineyard rows planted in uniform geometric grids.",
        "Permanently planted tree lines and aligned agricultural crop arrangements.",
        "Grid-aligned orchard trees and structured perennial cultivation plots."
    ],
    "Residential": [
        "Dense suburban layout featuring small roof structures and intersecting neighborhood streets.",
        "Housing blocks and residential properties with interconnected urban paths.",
        "Suburban neighborhood settlement with roof tops and local plot divisions."
    ],
    "River": [
        "A natural flowing water body channel curving through landmasses with visible banks.",
        "Winding river streams flowing through natural terrain boundaries.",
        "An elongated water channel cutting through the landscape."
    ],
    "SeaLake": [
        "A wide open body of water, coastal water body, or inland lake showing smooth surfaces.",
        "Large open water expanses with clear shoreline boundaries.",
        "A expansive static water body showing deep blue or green fluid textures."
    ]
}

# ==========================================
# 2. SIGReg LOSS (LeJEPA GAUSSIAN REGULARIZER)
# ==========================================
class SIGReg(nn.Module):
    def __init__(self, num_slices=1024):
        super().__init__()
        self.num_slices = num_slices

    def forward(self, z: torch.Tensor) -> torch.Tensor:
        B, D = z.shape
        if B < 2:
            return torch.tensor(0.0, device=z.device)

        slices = torch.randn(D, self.num_slices, device=z.device)
        slices = F.normalize(slices, p=2, dim=0)
        projected = torch.matmul(z, slices)

        diff_sq = (projected.unsqueeze(1) - projected.unsqueeze(0)) ** 2
        term1 = torch.exp(-0.5 * diff_sq).mean(dim=(0, 1))
        term2 = torch.exp(-0.25 * (projected ** 2)).mean(dim=0)

        c1 = torch.tensor(2.0, device=z.device).sqrt()
        c2 = 1.0 / torch.tensor(3.0, device=z.device).sqrt()

        ep_stat = term1 - c1 * term2 + c2
        return ep_stat.mean()

# ==========================================
# 3. TRANSFORM & MULTIMODAL DATASET
# ==========================================
class LeJEPAMultiCropTransform:
    def __init__(self, num_local_views=2): # 2 local views + 2 global views = 4 views total
        self.num_local_views = num_local_views
        self.global_transform = transforms.Compose([
            transforms.RandomResizedCrop(224, scale=(0.4, 1.0)),
            transforms.RandomHorizontalFlip(),
            transforms.ToTensor(),
            transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
        ])
        self.local_transform = transforms.Compose([
            transforms.RandomResizedCrop(96, scale=(0.08, 0.4)),
            transforms.RandomHorizontalFlip(),
            transforms.ToTensor(),
            transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
        ])

    def __call__(self, img):
        views = [self.global_transform(img), self.global_transform(img)]
        for _ in range(self.num_local_views):
            views.append(self.local_transform(img))
        return views


class MultimodalLeJEPADataset(Dataset):
    def __init__(self, root_dir="./data", max_samples=3000):
        self.transform = LeJEPAMultiCropTransform(num_local_views=2)
        self.eval_transform = transforms.Compose([
            transforms.Resize((224, 224)),
            transforms.ToTensor(),
            transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
        ])
        print("Loading EuroSAT Dataset...")
        self.eurosat = datasets.EuroSAT(root=root_dir, transform=None, download=True)
        self.classes = self.eurosat.classes

        max_samples = min(max_samples, len(self.eurosat))
        indices = list(range(len(self.eurosat)))
        random.seed(42)
        random.shuffle(indices)
        self.indices = indices[:max_samples]

        print("Loading text engine ('all-MiniLM-L6-v2')...")
        self.text_engine = SentenceTransformer('all-MiniLM-L6-v2')

    def __len__(self):
        return len(self.indices)

    def __getitem__(self, idx):
        real_idx = self.indices[idx]
        img, label_idx = self.eurosat[real_idx]
        class_name = self.classes[label_idx]

        variants = CLASS_DENSE_VARIANTS.get(class_name, [f"A satellite view of {class_name}."])
        caption = random.choice(variants)

        text_emb = self.text_engine.encode(caption, convert_to_tensor=True)
        text_emb = F.normalize(text_emb, dim=0)

        views = self.transform(img)
        eval_img = self.eval_transform(img)

        return views, text_emb, caption, eval_img

# ==========================================
# 4. MODEL WITH GRADIENT CHECKPOINTING
# ==========================================
class TextAlignedLeJEPA(nn.Module):
    def __init__(self, embed_dim=384):
        super().__init__()
        self.backbone = models.vit_b_16(weights=models.ViT_B_16_Weights.DEFAULT)
        self.backbone.heads = nn.Identity()

        self.projection_head = nn.Sequential(
            nn.Linear(768, 512),
            nn.LayerNorm(512),
            nn.GELU(),
            nn.Linear(512, embed_dim)
        )

        self.predictor = nn.Sequential(
            nn.Linear(embed_dim, embed_dim),
            nn.LayerNorm(embed_dim),
            nn.GELU(),
            nn.Linear(embed_dim, embed_dim)
        )

    def forward_encoder(self, x):
        if x.shape[-1] != 224:
            x = F.interpolate(x, size=(224, 224), mode='bilinear', align_corners=False)

        if self.training:
            features = checkpoint(self.backbone, x, use_reentrant=False)
        else:
            features = self.backbone(x)

        embeds = self.projection_head(features)
        return F.normalize(embeds, dim=-1)

    def forward(self, views):
        global_embeds = [self.forward_encoder(v) for v in views[:2]]
        local_embeds = [self.forward_encoder(v) for v in views[2:]]

        target_global = torch.stack(global_embeds, dim=0).mean(dim=0).detach()
        predictions = [self.predictor(l_emb) for l_emb in local_embeds]

        return target_global, global_embeds, predictions

# ==========================================
# 5. RETRIEVAL & EVALUATION HARNESS
# ==========================================
def evaluate_text_retrieval(query_text, dataset, model, device, top_k=3):
    model.eval()

    query_tensor = dataset.text_engine.encode(query_text, convert_to_tensor=True)
    query_vector = F.normalize(query_tensor, dim=0).to(device)

    eval_loader = DataLoader(dataset, batch_size=32, shuffle=False)

    all_image_embeds = []
    with torch.no_grad():
        for _, _, _, eval_imgs in eval_loader:
            eval_imgs = eval_imgs.to(device)
            embeds = model.forward_encoder(eval_imgs)
            all_image_embeds.append(embeds)

    all_image_embeds = torch.cat(all_image_embeds, dim=0)

    scores = torch.matmul(all_image_embeds, query_vector)
    top_scores, top_indices = torch.topk(scores, k=min(top_k, len(dataset)))

    print(f"\n==========================================")
    print(f"TEXT QUERY: '{query_text}'")
    print(f"==========================================")

    fig, axes = plt.subplots(1, top_k, figsize=(4 * top_k, 4.5))
    if top_k == 1:
        axes = [axes]

    mean = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1)
    std = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)

    for rank, (score, idx) in enumerate(zip(top_scores, top_indices), 1):
        idx_val = idx.item()
        score_val = score.item()

        _, _, caption, eval_img = dataset[idx_val]
        print(f"  Match {rank}: Index [{idx_val}] | Similarity Score: {score_val:.4f} | Label: {caption}")

        unnorm_img = eval_img * std + mean
        unnorm_img = torch.clamp(unnorm_img, 0, 1)
        img_np = unnorm_img.permute(1, 2, 0).numpy()

        axes[rank-1].imshow(img_np)
        axes[rank-1].set_title(f"Rank {rank}\nScore: {score_val:.3f}", fontsize=9)
        axes[rank-1].axis('off')

    plt.suptitle(f"LeJEPA Text Retrieval: '{query_text}'", fontsize=10, fontweight='bold')
    plt.tight_layout()
    plt.show()

# ==========================================
# 6. MAIN TRAINING LOOP
# ==========================================
if __name__ == "__main__":
    device = "cuda" if torch.cuda.is_available() else "cpu"
    print(f"Using compute device: {device}")

    batch_size = 2
    accum_steps = 16  # Effective batch size = 32

    train_ds = MultimodalLeJEPADataset(root_dir="./data", max_samples=3000)
    dataloader = DataLoader(train_ds, batch_size=batch_size, shuffle=True, num_workers=0, drop_last=True)

    model = TextAlignedLeJEPA(embed_dim=384).to(device)
    sigreg_loss_fn = SIGReg(num_slices=1024).to(device)

    scaler = torch.amp.GradScaler('cuda')

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=3e-4,
        weight_decay=0.05,
        fused=True if torch.cuda.is_available() else False
    )

    lambda_sigreg = 10.0
    lambda_text = 2.0
    epochs = 3

    print("\nStarting Memory-Optimized Multimodal LeJEPA Training...")
    model.train()

    for epoch in range(epochs):
        total_loss = 0.0
        optimizer.zero_grad(set_to_none=True)

        for step, (views, text_embeds, _, _) in enumerate(tqdm(dataloader, desc=f"Epoch {epoch+1}/{epochs}")):
            views = [v.to(device) for v in views]
            text_embeds = text_embeds.to(device)

            with torch.amp.autocast('cuda'):
                target_global, global_embeds, predictions = model(views)

                pred_loss = sum(F.mse_loss(pred, target_global) for pred in predictions) / len(predictions)
                sigreg_loss = sum(sigreg_loss_fn(g_emb) for g_emb in global_embeds) / len(global_embeds)
                text_align_loss = 1.0 - (target_global * text_embeds).sum(dim=-1).mean()

                loss = (pred_loss + (lambda_sigreg * sigreg_loss) + (lambda_text * text_align_loss)) / accum_steps

            scaler.scale(loss).backward()

            if (step + 1) % accum_steps == 0 or (step + 1) == len(dataloader):
                scaler.unscale_(optimizer)
                torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
                scaler.step(optimizer)
                scaler.update()
                optimizer.zero_grad(set_to_none=True)
                torch.cuda.empty_cache()

            total_loss += loss.item() * accum_steps

            del views, text_embeds, target_global, global_embeds, predictions, loss

        print(f"Epoch {epoch+1} | Combined Loss: {total_loss / len(dataloader):.4f}")

    # Run Zero-Shot Text Retrieval Evaluation
    evaluate_text_retrieval("paved transport highway roads with linear asphalt tracks", train_ds, model, device, top_k=3)
    evaluate_text_retrieval("dense canopied woodland area with heavy green vegetation", train_ds, model, device, top_k=3)
    evaluate_text_retrieval("large commercial buildings and concrete parking zones", train_ds, model, device, top_k=3)
```

```python
import os
import copy
import random
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import torchvision.datasets as datasets
import torchvision.transforms as transforms
import torchvision.models as models
import matplotlib.pyplot as plt
from tqdm import tqdm
from sentence_transformers import SentenceTransformer

# ==========================================
# 1. DATASET DEFINITION
# ==========================================
CLASS_DENSE_VARIANTS = {
    "AnnualCrop": [
        "Regularly plowed agricultural land with distinct parallel crop rows and soil patterns.",
        "Farmland with visible vegetation grids and tilled earth tracks.",
        "Cultivated crop fields displaying agricultural rows from an aerial perspective."
    ],
    "Forest": [
        "Dense canopied woodland area with heavy green vegetation cover and overlapping tree crowns.",
        "Thick forest terrain filled with natural organic distribution of green trees.",
        "A heavily wooded canopy zone showing dense vegetation patterns."
    ],
    "HerbaceousVegetation": [
        "Open fields covered with wild grass, low bushes, and uncultivated native terrain features.",
        "Grassy meadows and sprawling natural green pastures without human infrastructure.",
        "Low-lying herbaceous plants and open wild shrubbery terrain."
    ],
    "Highway": [
        "Paved transport highway roads with linear asphalt tracks and concrete lanes.",
        "Interstate asphalt corridors cutting across terrain with visible vehicular lanes.",
        "Linear transport highway infrastructure with dark asphalt strips."
    ],
    "Industrial": [
        "Large commercial buildings, metallic warehouse roofs, and concrete parking zones.",
        "Heavy industrial facilities with structured roofing and logistical yards.",
        "Commercial factory complexes and large metallic building structures."
    ],
    "Pasture": [
        "Open green livestock grazing fields, fenced grasslands, and agricultural meadows.",
        "Patchy green pastures used for grazing livestock with clear field boundaries.",
        "Open grassy fields designed for agricultural animal grazing."
    ],
    "PermanentCrop": [
        "Structured orchards, tree plantations, or vineyard rows planted in uniform geometric grids.",
        "Permanently planted tree lines and aligned agricultural crop arrangements.",
        "Grid-aligned orchard trees and structured perennial cultivation plots."
    ],
    "Residential": [
        "Dense suburban layout featuring small roof structures and intersecting neighborhood streets.",
        "Housing blocks and residential properties with interconnected urban paths.",
        "Suburban neighborhood settlement with roof tops and local plot divisions."
    ],
    "River": [
        "A natural flowing water body channel curving through landmasses with visible banks.",
        "Winding river streams flowing through natural terrain boundaries.",
        "An elongated water channel cutting through the landscape."
    ],
    "SeaLake": [
        "A wide open body of water, coastal water body, or inland lake showing smooth surfaces.",
        "Large open water expanses with clear shoreline boundaries.",
        "A expansive static water body showing deep blue or green fluid textures."
    ]
}

class EuroSATMultimodalDataset(Dataset):
    def __init__(self, root_dir="./data", max_samples=2000):
        self.transform = transforms.Compose([
            transforms.Resize((224, 224)),
            transforms.ToTensor(),
            transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
        ])
        print("Loading EuroSAT Dataset...")
        self.eurosat = datasets.EuroSAT(root=root_dir, transform=self.transform, download=True)
        self.classes = self.eurosat.classes

        max_samples = min(max_samples, len(self.eurosat))
        indices = list(range(len(self.eurosat)))
        random.seed(42)
        random.shuffle(indices)
        self.indices = indices[:max_samples]

    def __len__(self):
        return len(self.indices)

    def __getitem__(self, idx):
        real_idx = self.indices[idx]
        img, label_idx = self.eurosat[real_idx]
        class_name = self.classes[label_idx]

        variants = CLASS_DENSE_VARIANTS.get(class_name, [f"A satellite view of {class_name}."])
        caption = random.choice(variants)

        return img, caption, idx


# ==========================================
# 2. REAL VL-JEPA ARCHITECTURE COMPONENTS
# ==========================================

class VisionContextEncoder(nn.Module):
    """Encodes masked image context into latent representations."""
    def __init__(self, embed_dim=512):
        super().__init__()
        backbone = models.resnet34(weights=models.ResNet34_Weights.DEFAULT)
        self.stem = nn.Sequential(*list(backbone.children())[:-1])
        self.proj = nn.Linear(512, embed_dim)
        self.layer_norm = nn.LayerNorm(embed_dim)

    def forward(self, x):
        feat = self.stem(x).squeeze(-1).squeeze(-1)
        latents = self.layer_norm(self.proj(feat))
        return latents


class TextTargetEncoder(nn.Module):
    """Embeds textual modality targets into joint latent space."""
    def __init__(self, embed_dim=512):
        super().__init__()
        self.sentence_transformer = SentenceTransformer('all-MiniLM-L6-v2')
        # Freeze base sentence encoder parameters to act as stable semantic target
        for p in self.sentence_transformer.parameters():
            p.requires_grad = False

        self.proj = nn.Linear(384, embed_dim)
        self.layer_norm = nn.LayerNorm(embed_dim)

    def forward(self, text_list, device):
        with torch.no_grad():
            raw_embeds = self.sentence_transformer.encode(
                text_list, convert_to_tensor=True, device=device
            )
            # SentenceTransformer.encode() runs internally under
            # torch.inference_mode(), producing tensors that can never
            # re-enter an autograd graph. Clone to get a normal tensor.
            raw_embeds = raw_embeds.clone().detach()

        latents = self.layer_norm(self.proj(raw_embeds))
        return latents


class JEPAPredictor(nn.Module):
    """
    JEPA Predictor: Takes context embeddings and predicts target latent representation
    in the joint embedding space.
    """
    def __init__(self, embed_dim=512):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(embed_dim, embed_dim),
            nn.GELU(),
            nn.LayerNorm(embed_dim),
            nn.Linear(embed_dim, embed_dim)
        )

    def forward(self, context_latents):
        return self.net(context_latents)


class RealVLJEPA(nn.Module):
    """
    Joint-Embedding Predictive Architecture (VL-JEPA) wrapper with EMA updating
    target encoder and Latent Variance-Covariance Regularization (VICReg style).
    """
    def __init__(self, embed_dim=512, momentum=0.99):
        super().__init__()
        self.embed_dim = embed_dim
        self.momentum = momentum

        # Online / Context Encoder
        self.context_encoder = VisionContextEncoder(embed_dim=embed_dim)

        # EMA Target Encoder (Copy of context encoder for self-supervised vision stability)
        self.target_encoder = copy.deepcopy(self.context_encoder)
        for p in self.target_encoder.parameters():
            p.requires_grad = False

        # Cross-modal Text Encoder Target
        self.text_encoder = TextTargetEncoder(embed_dim=embed_dim)

        # Predictor Head
        self.predictor = JEPAPredictor(embed_dim=embed_dim)

    @torch.no_grad()
    def update_target_encoder(self):
        """EMA Momentum Update for Target Encoder."""
        for param_c, param_t in zip(self.context_encoder.parameters(), self.target_encoder.parameters()):
            param_t.data = param_t.data * self.momentum + param_c.data * (1.0 - self.momentum)

    def compute_jepa_loss(self, z_predicted, z_target):
        """
        JEPA Energy-based Latent Prediction Loss + Representation Variance Regularization
        """
        # 1. Prediction Distance (Smooth L1 in Latent Space)
        pred_loss = F.smooth_l1_loss(z_predicted, z_target)

        # 2. Variance Regularization (Prevents representation collapse)
        std_z = torch.sqrt(z_predicted.var(dim=0) + 1e-04)
        var_loss = torch.mean(F.relu(1.0 - std_z))

        # Combined Energy Minimization Loss
        return pred_loss + 0.5 * var_loss

    def forward(self, images, text_captions, device):
        # Generate context visual embedding
        z_context = self.context_encoder(images)

        # Predict target latent state
        z_pred = self.predictor(z_context)

        # Compute targets
        with torch.no_grad():
            z_target_vis = self.target_encoder(images)

        z_target_text = self.text_encoder(text_captions, device)

        # Calculate JEPA Losses
        loss_vis = self.compute_jepa_loss(z_pred, z_target_vis)
        loss_text = self.compute_jepa_loss(z_pred, z_target_text)

        return loss_vis + loss_text


# ==========================================
# 3. TEXT-TO-IMAGE RETRIEVAL ENGINE
# ==========================================
def retrieve_images_via_text_query(query_text, dataset, model, device, top_k=3):
    model.eval()

    # Project text query into joint latent space
    query_latent = model.text_encoder([query_text], device)
    query_latent = F.normalize(query_latent, dim=-1)

    dataloader = DataLoader(dataset, batch_size=64, shuffle=False)
    all_image_latents = []

    print(f"\nComputing latent representations across dataset for Query: '{query_text}'...")
    with torch.no_grad():
        for imgs, _, _ in dataloader:
            imgs = imgs.to(device)
            # Encode visual input into joint latent space
            img_latents = model.context_encoder(imgs)
            img_latents = F.normalize(img_latents, dim=-1)
            all_image_latents.append(img_latents)

    all_image_latents = torch.cat(all_image_latents, dim=0)  # [N, embed_dim]

    # Compute Cosine Similarity between query latent and visual latents
    similarities = torch.matmul(all_image_latents, query_latent.T).squeeze(-1)
    top_scores, top_indices = torch.topk(similarities, k=top_k)

    # Plot Visual Retrieval Results
    fig, axes = plt.subplots(1, top_k, figsize=(4 * top_k, 4.5))
    if top_k == 1:
        axes = [axes]

    mean = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1)
    std = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)

    print("\n--- RETRIEVAL RESULTS ---")
    for rank, (score, idx) in enumerate(zip(top_scores, top_indices), 1):
        idx_val = idx.item()
        img_tensor, original_caption, _ = dataset[idx_val]

        print(f"Rank {rank} | Score: {score.item():.4f} | Source Match Caption: {original_caption}")

        unnorm_img = img_tensor * std + mean
        unnorm_img = torch.clamp(unnorm_img, 0, 1)
        img_np = unnorm_img.permute(1, 2, 0).numpy()

        axes[rank - 1].imshow(img_np)
        axes[rank - 1].set_title(f"Rank {rank}\nSimilarity: {score.item():.3f}", fontsize=10)
        axes[rank - 1].axis('off')

    plt.suptitle(f"Real VL-JEPA Retrieval Query: '{query_text}'", fontsize=12, fontweight='bold')
    plt.tight_layout()
    plt.show()


# ==========================================
# 4. CHECKPOINT HELPERS
# ==========================================
CHECKPOINT_DIR = "./checkpoints"
CHECKPOINT_PATH = os.path.join(CHECKPOINT_DIR, "vljepa_last.pt")


def save_checkpoint(epoch, model, optimizer, avg_loss, path=CHECKPOINT_PATH):
    """
    Saves everything needed to resume exactly where training left off:
    - model weights (context encoder, target encoder, predictor, text proj)
    - optimizer state (momentum buffers etc.)
    - which epoch just finished
    - the running loss, just for logging continuity
    A temp file + os.replace is used so a crash mid-write can never
    corrupt the last good checkpoint.
    """
    os.makedirs(CHECKPOINT_DIR, exist_ok=True)
    tmp_path = path + ".tmp"
    torch.save({
        "epoch": epoch,
        "model_state_dict": model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "avg_loss": avg_loss,
    }, tmp_path)
    os.replace(tmp_path, path)  # atomic on POSIX + Windows
    print(f"[checkpoint] saved after epoch {epoch} -> {path}")


def load_checkpoint(model, optimizer, device, path=CHECKPOINT_PATH):
    """
    Returns the epoch to resume FROM (i.e. next epoch to run).
    If no checkpoint exists, returns 0 and leaves model/optimizer untouched.
    """
    if not os.path.exists(path):
        print("[checkpoint] none found, starting fresh from epoch 0")
        return 0

    print(f"[checkpoint] found {path}, loading...")
    ckpt = torch.load(path, map_location=device)
    model.load_state_dict(ckpt["model_state_dict"])
    optimizer.load_state_dict(ckpt["optimizer_state_dict"])
    last_epoch = ckpt["epoch"]
    print(f"[checkpoint] resuming after epoch {last_epoch} "
          f"(last avg loss: {ckpt.get('avg_loss', 'n/a')})")
    return last_epoch + 1  # next epoch to run


# ==========================================
# 5. TRAINING LOOP
# ==========================================
if __name__ == "__main__":
    device = "cuda" if torch.cuda.is_available() else "cpu"
    print(f"Executing on compute device: {device}")

    # Dataset & Dataloader
    dataset = EuroSATMultimodalDataset(root_dir="./data", max_samples=2000)
    train_loader = DataLoader(dataset, batch_size=32, shuffle=True, drop_last=True)

    # Instantiate VL-JEPA Model
    vljepa = RealVLJEPA(embed_dim=512, momentum=0.99).to(device)

    # Trainable parameters: Context Encoder + Predictor + Text Projection
    optimizer = torch.optim.AdamW(
        list(vljepa.context_encoder.parameters()) +
        list(vljepa.predictor.parameters()) +
        list(vljepa.text_encoder.proj.parameters()),
        lr=3e-4, weight_decay=1e-4
    )

    epochs = 3

    # Resume from a checkpoint if one exists (handles crashes / restarts)
    start_epoch = load_checkpoint(vljepa, optimizer, device)

    if start_epoch >= epochs:
        print(f"[checkpoint] already completed {start_epoch} epochs (target={epochs}); skipping training.")
    else:
        print("\nStarting Authentic VL-JEPA Training Loop...")

        for epoch in range(start_epoch, epochs):
            vljepa.train()
            running_loss = 0.0

            try:
                for imgs, captions, _ in tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs}"):
                    imgs = imgs.to(device)

                    optimizer.zero_grad()

                    # Predict latent representations across joint visual & text modalities
                    loss = vljepa(imgs, captions, device)

                    loss.backward()
                    torch.nn.utils.clip_grad_norm_(vljepa.parameters(), max_norm=1.0)
                    optimizer.step()

                    # Update target momentum encoder
                    vljepa.update_target_encoder()

                    running_loss += loss.item()
            except (KeyboardInterrupt, RuntimeError) as e:
                # Save whatever progress we made THIS epoch's worth of
                # updates before crashing/interrupting, tagged as the
                # previous completed epoch so it resumes at the same
                # epoch rather than skipping it half-done.
                print(f"\n[!] Training interrupted during epoch {epoch+1}: {e}")
                save_checkpoint(epoch - 1, vljepa, optimizer, running_loss, path=CHECKPOINT_PATH)
                raise

            avg_loss = running_loss / len(train_loader)
            print(f"Epoch [{epoch+1}/{epochs}] | Latent Prediction Loss: {avg_loss:.4f}")

            # Save checkpoint after every completed epoch
            save_checkpoint(epoch, vljepa, optimizer, avg_loss, path=CHECKPOINT_PATH)

    # ==========================================
    # 6. EXECUTE TEXT-TO-IMAGE RETRIEVAL
    # ==========================================
    retrieve_images_via_text_query(
        query_text="Paved transport highway roads with linear asphalt tracks",
        dataset=dataset,
        model=vljepa,
        device=device,
        top_k=3
    )

    retrieve_images_via_text_query(
        query_text="Dense canopied woodland area with heavy green vegetation",
        dataset=dataset,
        model=vljepa,
        device=device,
        top_k=3
    )
```
